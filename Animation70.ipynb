```python
import lightkurve as lk
import matplotlib.pyplot as plt
```

```python
def getData(ra:float, dec:float, sectorno:int):
    location = str(ra) + ", " + str(dec)
    search_result = lk.search_tesscut(location, sector=sectorno)
    tpf = search_result.download(cutout_size=60)
    return tpf
    
#getData(356, -3, 70)
```

```python
def tpfDraw(tpf, min_cadence, max_cadence, folder):
    i = min_cadence
    while i < max_cadence:
        tpf.plot(cadenceno = i)
        %matplotlib inline
        imagename = folder + '\Sector70_' + str(i) + '.png'
        plt.savefig(imagename)
        i += 500
    tpf.plot(cadenceno = max_cadence)
    %matplotlib inline
    plt.savefig(folder + '\Sector70_' + str(max_cadence) + '.png')

#tpfDraw(getData(356.7, -2.9, 70), 0, 4000)
```
