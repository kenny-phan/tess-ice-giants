```python
import lightkurve as lk
```

```python
search_result = lk.search_tesscut('356, -3', sector=70)
search_result
```

```python
tpf = search_result.download(cutout_size=60)
```

```python
target_mask = tpf.create_threshold_mask(threshold=15, reference_pixel='center')
n_target_pixels = target_mask.sum()
n_target_pixels
```

```python
tpf.plot()
%matplotlib inline
```

```python
help(tpf.plot)
```

```python
%matplotlib inline
tpf.plot(cadenceno = 10000);
```

```python
%matplotlib inline
tpf.plot(cadenceno = 10500, aperture_mask=target_mask, mask_color='k');
```

```python
%matplotlib inline
tpf.plot(cadenceno = 10715);
```
